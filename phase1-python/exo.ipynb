```python
def is_palindrome(s : str) -> bool:
    s = s.lower()
    s = s.replace(" ", "")
    l = s[::-1]
    return l == s 
print (is_palindrome("radar"))
print (is_palindrome("true"))
print(is_palindrome("Race car"))
```

```python
def count_vowels(s : str) :
    j = 0
    for letter in s:
        if letter in "aeuioy": 
            j += 1
    return j
print (count_vowels("abdeeeee"))
    
```

```python
def average(*nums) :
    if len(nums)==0 : 
        return 0
    else :
        return sum(nums)/len(nums)
print(average(1 , 2 , 3))
print(average())
```

```python
def clamp(x, low=0 , high=100):
    if x < 0 :
        x = 0
        return x
    elif x > 100 :
        x = 100
        return x
    else:
        return x 
print (clamp(150))
print (clamp(-52 , 50 , 101))
```

```python
def describe(**info):
    for k , v in info.items():
        print(f"{k.capitalize()}:{v}") 
describe(name="ada", age=36 , money=100000)       
```
